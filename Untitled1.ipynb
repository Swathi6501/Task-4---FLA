```python
import time
import multiprocessing
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def normal_program():
    total = 0
    for i in range(5):
        total += i
    return total


def infinite_loop():
    while True:
        pass


def conditional_loop():
    count = 0
    while count < 5:
        count += 1
    return count


def countdown():
    n = 5
    while n > 0:
        n -= 1
    return n


print("Normal Program Output:", normal_program())
print("Conditional Loop Output:", conditional_loop())
print("Countdown Output:", countdown())
```

```python
def run_with_timeout(function, timeout=2):
    process = multiprocessing.Process(target=function)
    process.start()
    process.join(timeout)

    if process.is_alive():
        process.terminate()
        process.join()
        return "Possible Infinite Loop"

    return "Terminated"


print("Normal Program:", run_with_timeout(normal_program))
print("Infinite Loop:", run_with_timeout(infinite_loop))
print("Conditional Loop:", run_with_timeout(conditional_loop))
print("Countdown:", run_with_timeout(countdown))
```

```python
programs = {
    "Normal Program": normal_program,
    "Infinite Loop": infinite_loop,
    "Conditional Loop": conditional_loop,
    "Countdown": countdown
}

results = []

for name, function in programs.items():
    result = run_with_timeout(function)
    results.append([name, result])

df = pd.DataFrame(results, columns=["Program", "Execution Status"])
print(df)
```

```python
colors = [
    "green" if status == "Terminated" else "red"
    for status in df["Execution Status"]
]

plt.figure(figsize=(9, 5))
plt.bar(df["Program"], [1] * len(df), color=colors)

plt.title("Program Termination Analysis")
plt.ylabel("Execution Status")
plt.xticks(rotation=20)
plt.yticks([])
plt.tight_layout()
plt.show()
```

```python
import ast

def detect_infinite_loop(code):
    tree = ast.parse(code)

    for node in ast.walk(tree):
        if isinstance(node, ast.While):
            if isinstance(node.test, ast.Constant) and node.test.value is True:
                has_break = any(
                    isinstance(child, ast.Break)
                    for child in ast.walk(node)
                )

                if not has_break:
                    return "Possible Infinite Loop"

                return "Needs Further Analysis"

    return "No Obvious Infinite Loop Detected"


test_programs = {
    "Program 1": """
while True:
    print("Running")
""",
    "Program 2": """
i = 0
while i < 5:
    i += 1
""",
    "Program 3": """
while True:
    break
"""
}

for name, code in test_programs.items():
    print(name, ":", detect_infinite_loop(code))
```

```python
def simulate_loop(start, limit, step, max_iterations=100):
    count = start
    iterations = 0

    while count < limit:
        if iterations >= max_iterations:
            return "Possible Infinite Loop", iterations

        count += step
        iterations += 1

    return "Terminated", iterations


test_cases = [
    ("Finite Loop", 0, 10, 1),
    ("Infinite Loop", 0, 10, 0),
    ("Another Finite Loop", 5, 20, 2)
]

results = []

for name, start, limit, step in test_cases:
    status, iterations = simulate_loop(start, limit, step)

    results.append([name, status, iterations])

df_simulation = pd.DataFrame(
    results,
    columns=["Program", "Execution Status", "Iterations"]
)

print(df_simulation.to_string(index=False))
```

```python
colors = [
    "green" if status == "Terminated" else "red"
    for status in df_simulation["Execution Status"]
]

plt.figure(figsize=(9, 5))

plt.bar(
    df_simulation["Program"],
    df_simulation["Iterations"],
    color=colors,
    edgecolor="black"
)

plt.title("Bounded Loop Simulation")
plt.xlabel("Program")
plt.ylabel("Number of Iterations")
plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Test different loop conditions

def test_loop_conditions():
    test_values = [0, 1, 5, 10, 20]

    results = []

    for value in test_values:
        count = 0

        while count < value:
            count += 1

        results.append({
            "Input": value,
            "Final Count": count,
            "Status": "Terminated"
        })

    return pd.DataFrame(results)


df_conditions = test_loop_conditions()
print(df_conditions.to_string(index=False))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df_conditions["Input"],
    df_conditions["Final Count"],
    marker="o",
    linewidth=2
)

plt.title("Loop Execution for Different Input Values")
plt.xlabel("Input Value")
plt.ylabel("Number of Iterations")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Simple Turing Machine Simulator

def run_turing_machine(tape_input, transitions, start_state="q0",
                       halt_state="HALT", max_steps=20):

    tape = {i: symbol for i, symbol in enumerate(tape_input)}
    head = 0
    state = start_state
    steps = 0

    while state != halt_state and steps < max_steps:
        symbol = tape.get(head, "_")

        if (state, symbol) not in transitions:
            break

        new_state, write_symbol, move = transitions[(state, symbol)]

        tape[head] = write_symbol
        state = new_state
        head += move
        steps += 1

    if state == halt_state:
        return "Machine Halted", steps

    return "Did Not Halt Within Step Limit", steps


# Machine 1: Moves right and halts at the blank symbol
halting_machine = {
    ("q0", "1"): ("q0", "1", 1),
    ("q0", "_"): ("HALT", "_", 0)
}

# Machine 2: Continues moving right forever
non_halting_machine = {
    ("q0", "1"): ("q0", "1", 1),
    ("q0", "_"): ("q0", "_", 1)
}

print("Halting Machine:",
      run_turing_machine("111", halting_machine))

print("Non-Halting Machine:",
      run_turing_machine("111", non_halting_machine))
```

```python
# Compare Turing Machine execution for different step limits

step_limits = [5, 10, 20, 50, 100]

halting_results = []
non_halting_results = []

for limit in step_limits:
    halting_results.append(
        run_turing_machine(
            "111", halting_machine, max_steps=limit
        )[1]
    )

    non_halting_results.append(
        run_turing_machine(
            "111", non_halting_machine, max_steps=limit
        )[1]
    )

plt.figure(figsize=(9, 5))

plt.plot(step_limits, halting_results, marker="o",
         label="Halting Machine")

plt.plot(step_limits, non_halting_results, marker="s",
         label="Non-Halting Machine")

plt.xlabel("Maximum Allowed Steps")
plt.ylabel("Steps Executed")
plt.title("Turing Machine Execution Comparison")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
